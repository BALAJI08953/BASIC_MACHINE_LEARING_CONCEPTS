```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
from sklearn.datasets import load_iris
data=load_iris()
```

```python
df=sns.load_dataset('iris')
```

```python
df
```

```python
x=df.iloc[:,:-1]
y=data.target
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.35,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
trans=DecisionTreeClassifier()
trans.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(trans,filled=True)
```

```python
#pre-prunning in pratical for decision tree
params={
    'criterion':["gini","entropy","log_loss"],
    "splitter":['best','random'],
    "max_depth":[1,2,3,4,5],
    "max_features":['auto','sqrt','log2']


}
```

```python
from sklearn.model_selection import GridSearchCV
trans=DecisionTreeClassifier(max_depth=2)
cv=GridSearchCV(trans,param_grid=params,cv=5,scoring='accuracy')
```

```python
cv.fit(x_train,y_train)
```

```python
cv.best_params_
```

```python
cv.best_index_
```

```python
y_predting=cv.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
score=accuracy_score(y_predting,y_test)
```

```python
print(score)
```

```python
classification_report(y_predting,y_test)
```
