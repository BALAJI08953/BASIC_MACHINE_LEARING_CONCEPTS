```python
#THIS IS AN EXAMPLE OF LINEAR RGESSION AND BADSIC FUNDAMENTALS OF THE MACHINE LEARNING AND DATA SCIENCE TO PREDICT THE VALUES AND THE RESLUTS FOR THE MODEL
xs=[1,2,3,4,5,6,7,8]
ys=[2.9,3.4,4.9,4.7,6.2,6.9,7.3,8.6]
w,b=0.0,0.0
lr=0.01
for epoch in range(1000):
    dw,db=0.0,0.0
    for x,y in zip(xs,ys):
        pred=w*x+b
        err=pred-y
        dw+=2*err*x/len(xs)
        db+=2*err/len(xs)
    w-=lr*dw
    b-=lr*db
if epoch % 100 == 0:
    print("Epoch:", epoch, "w:", w, "b:", b)

print("\nFinal values:")
print("w =", w)
print("b =", b)


print("\nPredictions:")

for x in xs:
    pred = w*x + b
    print("x =", x, "Predicted =", pred)
```
